```python
from transformers import GPT2LMHeadModel
```

```python
model_hf = GPT2LMHeadModel.from_pretrained("gpt2") #124M
sd_hf = model_hf.state_dict()  # return raw tensors

for k,v in sd_hf.items():
  print(k, v.shape)


# transformer.wte.weight torch.Size([50257, 768])  : wte = weight for token embeddings  , 50257: num of tokens in gpt2 vocab, 768: embedding dimension
# wpe = weight for position embedding   .... 1024:  maximum seq len , 768 : embedding dim
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
plt.imshow(sd_hf["transformer.wpe.weight"],cmap="gray")     
```

```python
plt.plot(sd_hf["transformer.wpe.weight"][:,150])
plt.plot(sd_hf["transformer.wpe.weight"][:,200])
plt.plot(sd_hf["transformer.wpe.weight"][:,250])
```
