```python
## • Unicode is a universal catalog or map. It assigns a unique, abstract number called a code point (usually written like U+0041) to every single character, letter, symbol, and emoji across almost every language in the world. Unicode tells you what number represents a character, but it doesn't specify how to write those numbers as binary data (0s and 1s) on a hard drive.
# • UTF-8 is a specific translation algorithm. It takes those abstract Unicode numbers and translates them into actual bytes so a computer can save, read, or transmit them.


# Because UTF-8 uses a variable-length system, it changes how many bytes it uses depending on the character being written:
# • 1 Byte: Standard English letters and numbers (the basic ASCII set) take just one byte (e.g., A is 01000001).
# • 2-3 Bytes: Middle Eastern, European, and Asian scripts take two to three bytes.
# • 4 Bytes: Complex symbols, rare historical characters, and emojis (like 😂) take four full bytes.
```

```python
"آپ کیسے ہیں؟ 👋 (How are you)".encode('utf-8')
```
