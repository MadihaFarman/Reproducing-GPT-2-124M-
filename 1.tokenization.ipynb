```python
## • Unicode is a universal catalog or map. It assigns a unique, abstract number called a code point (usually written like U+0041) to every single character, letter, symbol, and emoji across almost every language in the world. Unicode tells you what number represents a character, but it doesn't specify how to write those numbers as binary data (0s and 1s) on a hard drive.
# • UTF-8 is a specific translation algorithm. It takes those abstract Unicode numbers and translates them into actual bytes so a computer can save, read, or transmit them.


# Because UTF-8 uses a variable-length system, it changes how many bytes it uses depending on the character being written:
# • 1 Byte: Standard English letters and numbers (the basic ASCII set) take just one byte (e.g., A is 01000001).
# • 2-3 Bytes: Middle Eastern, European, and Asian scripts take two to three bytes.
# • 4 Bytes: Complex symbols, rare historical characters, and emojis (like 😂) take four full bytes.
```

```python
## Most modern Large Language Models (LLMs) do actually use UTF-8 as their foundational byte base (via Byte Pair Encoding or BPE), rather than throwing it out completely. However, they don't use raw UTF-8 characters as individual tokens for several critical reasons:

# 1. Context Window Bottlenecks and Sequence Lengths

# • Exploding token counts: If you treated every single Unicode character or UTF-8 byte as an individual token, a standard sentence would stretch into a massive sequence of tokens.
# • Memory and compute limits: LLMs have fixed context windows (e.g., 8k, 128k, or 1M tokens). Using raw character- or byte-level separation would waste precious context space on simple, repetitive words and letters
```

## BPE (byte pair encoding)

The original BPE algorithm operates by iteratively replacing the most common contiguous sequences of characters in a target text with unused 'placeholder' bytes. The iteration ends when no sequences can be found, leaving the target text effectively compressed.

#### Example

```python
text = "اسلام آباد(اردوپوائنٹ اخبارتازہ ترین-انٹرنیشنل پریس ایجنسی۔10 اکتوبر ۔2026 ) خیبر پختونخوا میں ایمرجنسی لگانے پر اتفاق ہوگیا ہے، ایمرجنسی کے نفاذ کی کل وفاقی کابینہ سے منظوری لیے جانے کا امکان ہے نجی ٹی وی نے ذرائع کے حوالے سے دعوی کیا ہے کہ ایمرجنسی کے نفاذ کی توثیق کے لیے پارلیمنٹ کے دونوں ایوانوں کے الگ الگ اجلاس طلب کیے جائیں گے، اجلاس میں منگل 13 اکتوبر کو قومی اسمبلی اور سینیٹ کے الگ الگ اجلاس بلانے پر اتفاق کیا گیا. ذرائع کا کہنا ہے کہ وفاقی کابینہ کا آجلاس کل صبح ساڑھے 9 بجے طلب کیا گیا ہے اس سے قبل خیبر پختونخوا کی صورتحال پر صدر مملکت آصف علی زرداری کی زیر صدارت ایوان صدر میں ہونے والے اہم اجلاس میں خیبر پختونخوا میں ایمرجنسی کے نفاذ پر غور کیا 👋 گیا."

tokens = text.encode('utf-8')   # raw bytes 
tokens = list(map(int,tokens))  # in order to easy to work with, map(convert) each token to int and create a list of it.
print("**********")
print(text)
print("length: ",len(text))
print("**********")
print(tokens)
print("length: ",len(tokens))
```
